**Research Question, Objectives and hypothesis**
Research Question: Do primary breast tumour and adjacent normal tissues exhibit statistically significant differential expression in key DNA-repair and metabolic pathway genes, and can these expression profiles accurately classify tissue phenotype using machine learning algorithms?    Project Objectives: Data Acquisition & Preprocessing: Acquire raw/count-level RNA-seq gene expression data from public repositories and perform data cleaning, log-transformation, and quality control checks.    Differential Expression Analysis: Perform differential gene expression analysis with False Discovery Rate (FDR) p-value adjustments to isolate significantly dysregulated genes.    Biological Pathway & Network Analysis: Map top differentially expressed genes onto KEGG/Reactome biological pathways and reconstruct protein interaction sub-networks to identify key functional hubs using NetworkX.    Machine Learning & Evaluation: Build and evaluate supervised machine learning classifiers (such as Random Forest and SVM) to predict tumour versus normal samples based on gene expression features


```python
import pandas as pd

# Load authoritative dataset and metadata directly from GitHub repository
base_url = "https://raw.githubusercontent.com/NyamburaWanjohi/Bioinformatics_project/main/data"

expr_df = pd.read_csv(f"{base_url}/expression_matrix.csv", index_col=0)
metadata_df = pd.read_csv(f"{base_url}/metadata.csv")

# Verify shapes and print required sample-count evidence table
print(f"Expression Matrix Shape (Genes x Samples): {expr_df.shape}")
print(f"Metadata Shape (Samples x Attributes): {metadata_df.shape}")
print("\nSample-Count Table:")
print(metadata_df['condition'].value_counts().rename_axis('Tissue Type').reset_index(name='Sample Count'))
print(f"\nUnique Patient Pairs: {metadata_df['patient_id'].nunique()}")
```
